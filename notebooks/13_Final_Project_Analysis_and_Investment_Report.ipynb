## 2. Methodology at a glance

- Five large Indian equities were used as the stock universe.
- Adjusted prices were used in the factor and optimization notebooks.
- Factor signals were formed from historical data and executed on a subsequent trading day in the factor backtests.
- The fixed-weight optimizer used a chronological training/test split: training through 2023-12-29 and test from 2024-01-01 through 2025-12-31.
- The walk-forward experiment used a trailing training window and quarterly-ish rebalancing, with 12 scheduled rebalance dates.
- The backtesting notebooks model transaction costs. Their results depend on the stated cost assumption and implementation choices.

These are historical simulations. They do not establish that the strategies will perform similarly in the future.

## 3. Factor strategy performance

Notebook 9 evaluated equal-weight, momentum, and low-volatility strategies alongside the NIFTY 50 reference over a common window from **2021-02-01 to 2025-12-31**, with 1,216 daily observations.

The table below transcribes the saved Notebook 9 output. Percentages and ratios are presented as reported by that notebook.

```python
import pandas as pd
from IPython.display import display

factor_results = pd.DataFrame({
    "Cumulative Return": ["75.29%", "69.17%", "113.90%", "91.64%"],
    "CAGR": ["12.33%", "11.51%", "17.07%", "14.43%"],
    "Annualized Volatility": ["15.33%", "17.16%", "17.62%", "13.94%"],
    "Sharpe Ratio": [0.836, 0.721, 0.983, 1.037],
    "Sortino Ratio": [1.233, 1.047, 1.479, 1.492],
    "Maximum Drawdown": ["-15.96%", "-22.25%", "-25.38%", "-17.23%"],
    "Positive Days": ["51.81%", "52.30%", "50.82%", "53.70%"]
}, index=["Equal-weight", "Momentum", "Low-volatility", "NIFTY 50"])

display(factor_results)
```

### Interpretation

- **Low-volatility had the highest cumulative return and CAGR among the three strategies** in this experiment, and its Sharpe ratio (0.983) exceeded those of equal-weight and momentum.
- **NIFTY 50 had the highest Sharpe ratio in the comparison** (1.037) and the lowest annualized volatility (13.94%).
- Low-volatility's maximum drawdown was -25.38%, larger in magnitude than the benchmark's -17.23%. Higher return did not mean lower downside risk in this sample.
- Momentum trailed equal-weight on CAGR and Sharpe ratio in this particular test.

Do not generalize this ranking beyond the selected universe, period, and strategy definitions.

## 4. Fixed-weight portfolio optimization: held-out test

## 5. Robustness checks

Notebook 11 varied position caps, risk-free-rate assumptions, covariance estimation, and the expected-return estimation window. The saved results showed that optimized weights can change materially when assumptions change.

Examples from the executed experiment:
- Under Ledoit-Wolf covariance shrinkage, the TCS weight changed from approximately **44.58% to 41.66%** in the compared maximum-Sharpe case.
- Under expected-return window sensitivity, the maximum-Sharpe TCS allocation shifted from approximately **10.25% using the full training period to the 50% cap using the recent-year estimate**.
- The maximum-Sharpe allocation to Infosys shifted from approximately **50% to zero** across those return-window estimates.

These changes are evidence of **estimation sensitivity**. In particular, historical expected returns are noisy, and a mathematically optimal portfolio can be unstable when small changes in inputs lead to large changes in weights.

The five-stock universe is too narrow to claim that these findings generalize to the Indian equity market as a whole.

## 6. Walk-forward portfolio backtesting

Notebook 12 used a rolling training window, approximately quarterly rebalancing, and a default transaction-cost assumption of 10 basis points per one-way turnover. The executed notebook generated **12 scheduled rebalance dates**, with all 12 optimization dates reported as successful or minimally handled.

The corrected return calculation leaves returns missing before the first portfolio execution date. Those pre-execution dates must not be treated as zero-return observations.

```python
walk_forward_results = pd.DataFrame({
    "Cumulative Return": ["32.10%", "29.30%", "27.63%"],
    "Annualized Return": ["10.10%", "9.29%", "8.80%"],
    "Sharpe Ratio": [0.336, 0.283, 0.229],
    "Maximum Drawdown": ["-15.22%", "-13.29%", "-18.56%"]
}, index=["Equal Weight", "Rolling Minimum Variance", "Rolling Maximum Sharpe"])

display(walk_forward_results)
```

### Interpretation

- Equal-weight led the three walk-forward strategies on cumulative return, annualized return, and Sharpe ratio in this saved run.
- Rolling minimum variance had the least severe maximum drawdown of the three.
- Rolling maximum Sharpe did not outperform equal-weight in this period. This is a useful empirical result: optimization is not guaranteed to improve realized performance.
- The walk-forward period starts only after sufficient training history was available, so it is not directly comparable with the full-period factor experiment.

## 8. Limitations

1. **Narrow universe:** only five large-cap Indian stocks were studied.
2. **Sample dependence:** the conclusions may depend strongly on the chosen start/end dates and market regime.
3. **Estimation error:** expected returns and covariance matrices are estimated from historical data and are uncertain.
4. **Implementation simplifications:** transaction costs are modeled rather than taken from actual trades; taxes, market impact, liquidity constraints, and slippage may not be fully captured.
5. **Benchmark consistency:** benchmark returns must be aligned to the exact same dates as each strategy.
6. **No guarantee of future performance:** historical backtests are not forecasts.
7. **Potential data-provider changes:** downloaded adjusted prices may change as the data provider revises historical data.
